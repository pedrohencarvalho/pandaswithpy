```python
import pandas as pd
import numpy as np

#dataframe e a tabela estrutura dentro do frame, base de dados do panda (df) // forma de carregar e visualizar
df_vendas = pd.read_csv("vendas_tech.csv", low_memory=False)
df_gerentes = pd.read_excel("gerentes_lojas.xlsx")
display(df_vendas)
display(df_gerentes)
```

```python
#inspecao
display(df_vendas)
#display(df_vendas.head(15)) #primeiras
#display(df_vendas.tail(15)) #ultimas
#display(df_vendas.sample(15)) #aleatorio
display(df_vendas.shape) #tamanho da tabela
display(df_vendas.columns) #colunas
display(list(df_vendas.columns)) #colunas, transformando em lista PY
display(df_vendas.info()) #pegar informaçoes, e verificar erros
display(df_vendas.describe()) #descrever a base de dados, descricao estatistica, numericos, medicas, valor miniomo, maximo, quartos
```

```python
#tratamendo de dados
#display(df_vendas["Loja"]) #acessar coluna, seleçao de coluna
#display(df_vendas[["Loja", "Cliente"]]) #acessar mais colunas, colocar dois couchetes ou setar antes numa variavel as colunas que quiser

#trabalhar com colunas
df_vendasv2 = df_vendas.drop(columns=["Data_Base"])#excluir uma coluna, duas tres, varias (criando a v2 para nao mexer na original)

#tratar valores nulos
#df_vendasv2 = df_vendasv2.dropna() #comando mais simples, exlui todas as linhas que tem 1 valor vazio
df_vendasv2["Loja"] = df_vendasv2["Loja"].fillna("Online") #editando apenas coluna loja, porem poderia colocar tudo em uma linha

#tipos de dados
df_vendasv2["Data"] = pd.to_datetime(df_vendasv2["Data"], format="%Y-%m-%d") #transforma o tipo da coluna para data do py **pesquisar tabela para usar o formato correto

#padronizaçao
df_vendasv2["Loja"] = df_vendasv2["Loja"].str.strip() #remove espaço duplicado, espaço vazio no inicio ou final do texto
df_vendasv2["Loja"] = df_vendasv2["Loja"].str.title() #padronizar o nome das lojas com a primeira letra maiuscula resto minusculo

df_gerentes["Loja"] = df_gerentes["Loja"].str.strip() #remove espaço duplicado, espaço vazio no inicio ou final do texto
df_gerentes["Loja"] = df_gerentes["Loja"].str.title() #padronizar o nome das lojas com a primeira letra maiuscula resto minusculo

#duplicatas
df_vendasv2 = df_vendasv2.drop_duplicates(subset=["ID_Pedido"], keep="first")#remove duplicadas, mantem o primeiro valor

display(df_vendasv2)
display(df_vendasv2.info())
```

```python
# criando novas colunas
#faturamento
df_vendasv2["Faturamento"] = df_vendasv2["Qtd"] * df_vendasv2["Preco_Unitario"] #adicionando uma nova coluna de Faturamento entre valor do produto e a quantidade

#forma de venda
df_vendasv2["Forma_de_Venda"] = np.where(df_vendasv2["Loja"] == "Online", "Online", "Presencial") #aplicando condicional como se fosse if, else, para adicionar a colona Forma de venda entre online e presencial

#regiao------------
display(df_vendasv2["Loja"].unique()) #usando funçao, coluna especifica ex(unique, mean,sum,count,value_counts)
#opcao1
#df_vendasv2["Regiao"] = np.where(df_vendasv2["Loja"]== "Porto alegre", "Sul")...
#opcao2
#sul = ["Curitiba", "Porto Alegre"]
#df_vendasv2["Regiao"] = np.where(df_vendasv2["Loja"].isin(sul), "Sul")
#opcao3
# def definir_regiao(loja): #criando uma funcao metodo complexo
#     dic_regioes = {
#     'São Paulo' : "Sudeste",
#     'Belo Horizonte' : "Sudeste",
#     'Online' : "Online",
#     'Rio De Janeiro' : "Sudeste",
#     'Salvador' : "Nordeste",
#     'Recife' : "Nordeste",
#     'Curitiba' : "Sul",
#     'Porto Alegre' : "Sul"
# }
#     return dic_regioes[loja]

#opcao4 - apenas com o dicionario ja basta
dic_regioes = {
    'São Paulo' : "Sudeste",
    'Belo Horizonte' : "Sudeste",
    'Online' : "Online",
    'Rio De Janeiro' : "Sudeste",
    'Salvador' : "Nordeste",
    'Recife' : "Nordeste",
    'Curitiba' : "Sul",
    'Porto Alegre' : "Sul"
}

df_vendasv2["Regiao"] = df_vendasv2["Loja"].map(dic_regioes)

display(df_vendasv2)
#display(df_vendasv2.info())
display(df_vendasv2.isna().sum()) #aplicando funcao na base inteira. em todas colunas, verificando se esta vazio e somar para verificar
```

```python
#analisar - filtrar

#ordenar a base por coluna
df_vendasv2 = df_vendasv2.sort_values(by=["Data", "Faturamento"]) ##se quiser mudar a ordem do maior pro menor anscendig=
df_vendasv2 = df_vendasv2.reset_index(drop=True) #reset no index para deixar os indices do 0 ao ultimo, e deletar o indice antigo, para nao criar outra coluna


#localizar algo especifico em uma linha e uma coloca # .loc: id do pedido - - pelo nome da coluna e do indice
id_pedido = 4
loja = df_vendasv2.loc[df_vendasv2["ID_Pedido"] == 4, "Loja"].values[0]
produto = df_vendasv2.loc[df_vendasv2["ID_Pedido"] == 4, "Produto"].values[0]
cliente = df_vendasv2.loc[df_vendasv2["ID_Pedido"] == 4, "Cliente"].values[0]
print(loja)
print(produto)
print(cliente)

#por posicao -> .iloc
id_pedido = 4
loja = df_vendasv2.iloc[3,2]
produto = df_vendasv2.iloc[3,3]
cliente = df_vendasv2.iloc[3,6]
print(loja, produto, cliente)


#condicional - linha toda da tabela separada
#df_id_pedido_4 = df_vendasv2[df_vendasv2["ID_Pedido"] == 4]

#exportar pedaços da base
#vendas de sp
df_vendas_sp = df_vendasv2[df_vendasv2["Loja"] == "São Paulo"]
#exportar pro chefe, mandar para alguem
df_vendas_sp.to_csv("Vendas_SP.csv", index=False)

#exportar vendas de 2024
df_vendas_2024 = df_vendasv2[df_vendasv2["Data"].dt.year == 2024]
df_vendas_2024.to_csv("Vendas_2024.csv", index=False)

#duplas condicoes
df_vendas_HDMI_SUL = df_vendasv2[(df_vendasv2["Produto"] == "Cabo HDMI") & (df_vendasv2["Regiao"] == "Sul")]

display(df_vendas_HDMI_SUL)
```

```python
#analises por agrupamento
#display(df_vendasv2)
#qual loja vendeu mais?
analise_lojas = df_vendasv2[["Loja", "Faturamento"]].groupby("Loja").sum() ##pegando o faturamento de cada loja e somando cada faturamento
analise_lojas = analise_lojas.sort_values(by="Faturamento", ascending=False) #ordenando do maior para o menor
analise_lojas = analise_lojas.reset_index()
analise_lojas["Faturamento"] = analise_lojas["Faturamento"].map("R${:,.2f}".format) ##formatando para real
display(analise_lojas)

#ranking de produtos que mais venderam no online
df_vendas_online = df_vendasv2[df_vendasv2["Loja"] == "Online"]
analise_produtos_online = df_vendas_online[["Produto", "Qtd"]].groupby("Produto").sum() ##pegando a quantidade de cada produto e somando qual vendeu mais
df_produtos_online = analise_produtos_online.sort_values(by="Qtd", ascending=False) #ordenando do maior para o menor
analise_produtos_online = analise_produtos_online.rename(columns={"Qtd":"Vendas Totais"}) #trocar o nome da coluna de uma tabela
display(analise_produtos_online)

#analise de ranking por loja e por produto
#quais produtos venderam mais em cada uma das lojas
analise_produtos_em_lojas = df_vendasv2[["Loja", "Produto", "Qtd"]].groupby(["Loja", "Produto"]). sum() ##pegando a quantidade de cada loja e produto e somando qual vendeu mais

#quais lojas mais venderam os produtos
analise_lojas_em_produtos = df_vendasv2[["Loja", "Produto", "Qtd"]].groupby(["Produto", "Loja"]). sum() ##pegando a quantidade de cada produto e loja e somando qual vendeu mais


##somente para este display fazer aparecer todos os dados da tabela
with pd.option_context("display.max_rows", None):
    display(analise_produtos_em_lojas)
    display(analise_lojas_em_produtos)
```

```python
display(df_vendasv2.head())

#quais gerentes bateram a meta em janeiro de 2023--------------
df_meta = df_vendasv2[(df_vendasv2["Data"].dt.year == 2023) & (df_vendasv2["Data"].dt.month == 1)]

df_meta = df_meta[["Loja", "Faturamento"]].groupby("Loja", as_index=False).sum()  #agrupamento por loja e somando o faturamento

#juntando duas bases de dados
#df_meta = pd.concat([tabela1, tabela2]) necessario ter colunas identicas, juntar uma na outra, so linhas diferentes

#existe tb how=""left, inner, outer, join, inner join, left join
df_meta = df_meta.merge(df_gerentes, on="Loja", how="left") # tem o mesmo nome de coluna em ambas tabelas, posso usar o on, se for nome diferente usar left on, right on
df_meta["Bateu Meta"] = np.where(df_meta["Faturamento"] >= df_meta["Meta_Mensal"], "Sim", "Nao")
display(df_meta)
```

```python
#faturamento por mes/ano
df_vendasv2["Mes-Ano"] = df_vendasv2["Data"].dt.to_period("M") #criando a coluna mes ano e nela sendo apenas o mes da coluna data
df_vendas_mes = df_vendasv2[["Mes-Ano", "Faturamento"]].groupby("Mes-Ano").sum()
#df_vendas_mes.plot() plotly
df_vendas_mes["Faturamento"] = df_vendas_mes["Faturamento"].map("R${:,.2f}".format) ##formatando para real
df_vendas_mes = df_vendas_mes.sort_values(by="Faturamento", ascending=False) #ordenando do maior para o menor
display(df_vendas_mes)
```
